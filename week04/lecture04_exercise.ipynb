```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE

# Task 1 — Scatter: life expectancy change over time

import plotly.express as px
import plotly.graph_objects as go

# Choose one continent (not Africa)
continent_choice = "Asia"

# Filter data for the selected continent and years
df_task1 = df[
    (df["continent"] == continent_choice) &
    (df["year"].isin([2002, 2007]))
].copy()

# Create scatter plot with two colours for the two years
fig = px.scatter(
    df_task1,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    color_discrete_map={2002: "blue", 2007: "red"},
    hover_name="country",
    log_x=True,
    title="GDP per Capita vs Life Expectancy in Asia (2002 and 2007)",
    labels={
        "gdpPercap": "GDP per Capita (log scale)",
        "lifeExp": "Life Expectancy (years)",
        "year": "Year"
    }
)

# Connect each country's 2002 and 2007 points to show change direction
for country in df_task1["country"].unique():
    country_data = df_task1[df_task1["country"] == country].sort_values("year")

    if len(country_data) == 2:
        fig.add_trace(
            go.Scatter(
                x=country_data["gdpPercap"],
                y=country_data["lifeExp"],
                mode="lines",
                line=dict(color="gray", width=1),
                opacity=0.5,
                showlegend=False,
                hoverinfo="skip"
            )
        )

fig.update_layout(
    template="plotly_white",
    legend_title_text="Year"
)

fig.show()
```

## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
# Task 2
# YOUR CODE HERE
# Task 2 — Bubble chart: tell a story

import plotly.express as px

# Select the 2007 data for all countries
df_2007 = df[df["year"] == 2007].copy()

# Choose one continent to highlight
continent_choice = "Asia"

# Create a grey colour for other continents and highlight Asia
df_2007["group"] = df_2007["continent"].apply(
    lambda continent: continent_choice
    if continent == continent_choice
    else "Other continents"
)

# Create the bubble chart
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="group",
    hover_name="country",
    log_x=True,
    size_max=55,
    color_discrete_map={
        "Asia": "orange",
        "Other continents": "lightgrey"
    },
    title="GDP per Capita vs Life Expectancy (2007)",
    labels={
        "gdpPercap": "GDP per Capita (log scale)",
        "lifeExp": "Life Expectancy (years)",
        "pop": "Population",
        "group": "Continent"
    }
)

# Add an annotation explaining Asia's story
fig.add_annotation(
    x=15000,
    y=72,
    text="Asia: countries vary widely in income<br>and life expectancy",
    showarrow=True,
    arrowhead=2,
    ax=70,
    ay=-60,
    bgcolor="white"
)

# Improve the chart layout
fig.update_layout(
    template="plotly_white",
    legend_title_text="Group"
)

fig.show()
```
